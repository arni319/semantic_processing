```python
import numpy as np
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF
import pandas as pd
import matplotlib.pyplot as plt
```

```python
reviews = []
for file in os.listdir(ROOT):               # Lists all files in the directory
    path = os.path.join(ROOT, file)         # Gets the file paths of file in directory
    if os.path.isfile(path):
        with open(path, 'r', encoding='utf-8') as fin:
            reviews.append(fin.read())
```

## Feature Extraction

```python
vect = TfidfVectorizer(stop_words='english')
X = vect.fit_transform(reviews)

pd.DataFrame(X.toarray(), columns=vect.get_feature_names_out())
```

## NMF Decomposition

```python
N_TOPICS = 20
nmf = NMF(n_components=N_TOPICS, max_iter=1000)
W = nmf.fit_transform(X)  # Document-topic matrix
H = nmf.components_       # Topic-term matrix
```

```python
# Top 10 words per topic

words = np.array(vect.get_feature_names_out())
topic_words = pd.DataFrame(np.zeros((N_TOPICS, 10)), index=[f'Topic {i + 1}' for i in range(N_TOPICS)],
                           columns=[f'Word {i + 1}' for i in range(10)]).astype(str)
for i in range(N_TOPICS):
    ix = H[i].argsort()[::-1][:10]
    topic_words.iloc[i] = words[ix]

topic_words
```

```python
# Create a topic mapping

topic_mapping = {
    'Topic 4': 'TV',
    'Topic 7': 'Sci-fi',
    'Topic 8': 'Comedy',
    'Topic 10': 'Animation',
    'Topic 11': 'Musical',
    'Topic 12': 'Book Adaptation',
    'Topic 13': 'Horror',
    'Topic 15': 'Martial Arts / Action',
    'Topic 16': 'Romance',
    'Topic 17': 'War',
    'Topic 18': 'Book Adaptations',
    'Topic 19': 'Game Adaptations'
}
```

```python
# Recall the document-topic matrix, W

W = pd.DataFrame(W, columns=[f'Topic {i + 1}' for i in range(N_TOPICS)])
W['max_topic'] = W.apply(lambda x: topic_mapping.get(x.idxmax()), axis=1)
Film_topics = W[pd.notnull(W['max_topic'])]
```

```python
Film_topics.columns
```

```python
# Film types by review topics
Film_topics['max_topic'].value_counts()
```

```python
topics = Film_topics['max_topic'].unique()
```

```python
review_samples = {}

# Random sampling 2 review-indices based on topics
for topic in topics:
    review_samples[topic] = Film_topics[Film_topics['max_topic'] == topic].sample(2).index.values

# Printing reviews of topic based sampled indices
for key, vals in review_samples.items():
    print(f'Topic: {key}')
    for idx in vals:
        print(reviews[idx])
        print()
    print('-'*100)
```
