```python
import re
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
sent = 'The quick brown fox jumps over lazy dog.'
```

```python
pattern = re.compile('Xx+|x+')
words = [tok.text for tok in nlp(sent) if pattern.match(tok.shape_)]
words
```

```python
cbow = []
for i in range(len(words)):
    if i == 0:
        cbow.append(([words[i+1]], words[i]))
    elif i == len(words)-1:
        cbow.append(([words[i-1]], words[i]))
    else:
        cbow.append(([words[i-1], words[i+1]], words[i]))

cbow
```

```python
sorted(words, key=lambda x: x[0], reverse= True)
```
